```python
"""
Example: Loading a Pretrained TempoVital-RT Model

This script demonstrates how to:
- Load a tokenizer and model from a saved TempoVital-RT checkpoint
- Verify PyTorch and Transformers versions
- Move the model to GPU (if available)
"""
```

```python
# ---------------------------------------------
# 1. Import Dependencies
# ---------------------------------------------
import torch
from transformers import BertTokenizer, BertModel

print("torch version:", torch.__version__)

import transformers
print("transformers version:", transformers.__version__)
```

```python
# ---------------------------------------------
# 2. Device Settings
# ---------------------------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

```python
# ---------------------------------------------
# 3. Load Pretrained Model & Tokenizer
# ---------------------------------------------
# Set your model path (example: 5-minute SpO2 model)
model_path = "MIMIC+EICU/SpO2/5min/Model/"

print(f"Loading tokenizer from: {model_path}")
tokenizer = BertTokenizer.from_pretrained(model_path)

print(f"Loading model from: {model_path}")
model = BertModel.from_pretrained(model_path).to(device)

print("Model and tokenizer loaded successfully!")
```
